```python
import pandas as pd
```

```python
data = {
    'age':[42,52,36,24,73],
    'preTestScore':[4,24,31,2,3],
    'postTestScore':[25,94,57,62,70]
}
df = pd.DataFrame(data)
df
```

```python
def isKAnonymized(df,k):
    for index,row in df.iterrows():
        query = '&'.join([f'{col} == {row[col]}' for col in df.columns])
        rows = df.query(query)
        if rows.shape[0] < k:
            return False
    return True
isKAnonymized(df,1)
isKAnonymized(df,2)
```
